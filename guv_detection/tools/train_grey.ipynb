**The following code will train the network**

````python
from ultralytics import YOLO

# Load a model

model = YOLO("yolo11n.yaml")  # build a new model from YAML

model = YOLO("yolo11n.pt")  # load a pretrained model (recommended for training)
model = YOLO("yolo11n.yaml").load("yolo11n.pt")  # build from YAML and transfer weights

# Train the model
results = model.train(data="data.yaml", epochs=100, imgsz=640)
````

**YOLO inference i.e. to use the trained model to analyze a video**

````python
from ultralytics import YOLO
import os

# Load a pretrained YOLO11n model
model = YOLO("runs/detect/train/weights/best.pt") # Give full address of the saved weight file

# Run inference on videos (some_name.avi) or image (some_name.png) with arguments
for i in os.listdir("test/images"):
  result = model.predict(f"test/images/{i}", save=True, imgsz=640, save_txt=True, show_labels=False, show_conf=False, line_width=2, conf=0.35, batch=1) # Give full address of the input video file
  print(result)
  exit()
````

**Once the images/videos are analyzed, the results are stored in runs/detect/predictX/ directory. X is an integer**
